```python
##1. remove and print in reverse insertion order using stacks

stack=[]
num=input("Enter a list of numbers: ").split()

for i in num:
    stack.append(i)

while stack:
    print(f"{stack.pop()}")
```

```python
##3. Support Ticket System

from collections import deque

q=deque()

tickets=int(input("How many tickets u want to enter: "))

for i in range(tickets):
    ticket={}
    ticket["id"]=len(q)+1
    ticket["customer_name"]=input("Enter customer name: ")
    ticket["issue"]=input("Issue: ")
    q.append(ticket)

print(f"Queue: {q}")

while q:
    ticket = q.popleft()

    print(f"\n\nProcessing ticket #{ticket['id']}")
    print(f"Customer name: {ticket['customer_name']}")
    print(f"Issue: {ticket['issue']}")

print(f"\n\nAll tickets have been processed.")
```

```python
##4. Priority Support Queue

from collections import deque

q = deque()

n = int(input("How many tickets? "))

for i in range(n):
    ticket["id"] = int(input("Ticket ID: "))
    ticket["customer_name"] = input("Customer name: ")
    ticket["issue"] = input("Issue: ")
    ticket["priority"] = int(input("Priority (1=Critical, 2=High, 3=Normal): "))

    q.append(ticket)

while q:

    highest_priority = 4

    for ticket in q:
        if ticket["priority"] < highest_priority:
            highest_priority = ticket["priority"]

    for i in range(len(q)):
        if q[i]["priority"] == highest_priority:
            ticket = q[i]
            del q[i]
            break

    print(f"\nProcessing Ticket #{ticket['id']}")
    print(f"Customer: {ticket['customer_name']}")
    print(f"Issue: {ticket['issue']}")
    print(f"Priority: {ticket['priority']}")
    print(f"Tickets remaining: {len(q)}")

print(f"\nAll tickets have been processed.")
```

```python
##5. Support Ticket System

from queue import Queue

q=Queue()

for i in range(5):
    ticket={}
    ticket["id"] = q.qsize()+1
    ticket["customer_name"] = input("Customer name: ")
    ticket["issue"] = input("Issue: ")

    q.put(ticket)


while not q.empty():
    ticket=q.get()

    print(f"\nProcessing Ticket #{ticket['id']}")
    print(f"Customer: {ticket['customer_name']}")
    print(f"Issue: {ticket['issue']}")
    q.task_done()

print(f"\nAll tickets have been processed.")
```

```python
##4. Priority Support Queue using heapq

import heapq

q = []

n = int(input("How many tickets? "))

for i in range(n):
    ticket={}
    ticket["id"] = int(input("Ticket ID: "))
    ticket["customer_name"] = input("Customer name: ")
    ticket["issue"] = input("Issue: ")
    priority = int(input("Priority (1=Critical, 2=High, 3=Normal): "))

    heapq.heappush(q,(priority,ticket))

while q:
    print(heapq.heappop(q))
print(f"\nAll tickets have been processed.")
```

```python
##5. Linear search

numbers=[12,34,45,67,78,67]
num=int(input("Enter a number: "))

flag=False
for i in range(len(numbers)):
    if numbers[i]==num:
        print(f"Number found at index: {i}")
        flag=True

if flag==False:
    print("Number not found!")
```

```python
##6. Binary Search

numbers = [3, 8, 12, 17, 21, 29, 35, 42, 50]

num=int(input("Enter a number to find: "))

found=False
left=0
right=len(numbers)-1

while left<=right:

    mid=(left+right)//2

    if num==numbers[mid]:
        print(f"Number found at index: {mid}")
        found=True
        break
    
    elif num<numbers[mid]:
         right=mid-1

    else:
         left=mid+1

if found==False:
     print("Number not found!")
```

```python
##7. Bubble Sort

numbers = [64, 34, 25, 12, 22, 11, 90]

for i in range(len(numbers)-1):
    for j in range(len(numbers)-1):
        if numbers[j]>numbers[j+1]:
            numbers[j],numbers[j+1]=numbers[j+1],numbers[j]

print(f"Sorted list in ascending order: {numbers}")
```

```python
##8. Bubble sort in descending order

numbers = [23, 7, 45, 12, 89, 34, 2]


for i in range(len(numbers)-1):
    for j in range(len(numbers)-1):
        if numbers[j]<numbers[j+1]:
            numbers[j],numbers[j+1]=numbers[j+1],numbers[j]

print(f"Sorted list in descending order: {numbers}")
```

```python
##9. Hash maps

class HashTable:
    def __init__(self,size):
        self.size=size
        self.hash_table=([] for _ in range(size))

    def set_val(self,key,val):
        hashed_key=hash(key)%self.size
        bucket=self.hash_table[hashed_key]

        for index, (record_key,_) in enumerate(bucket):
            if record_key==key:
                bucket[index]=(key,val)
                return

        bucket.append(self,key)

    def get_val(self, key):
        hashed_key = hash(key) % self.size
        bucket = self.hash_table[hashed_key]

        for record_key, record_val in bucket:
            if record_key == key:
                return record_val
        return "No record found"

    def delete_val(self, key):
        hashed_key = hash(key) % self.size
        bucket = self.hash_table[hashed_key]

        for index, (record_key, _) in enumerate(bucket):
            if record_key == key:
                bucket.pop(index)
                return

    def __str__(self):
        return "".join(str(bucket) for bucket in self.hash_table)
        
ht = HashTable(3)

ht.set_val('apple', 10)
ht.set_val('banana', 20)
ht.set_val('cherry', 30)

print("Hash Table:", ht)

print("Value for 'banana':", ht.get_val('banana'))
print("Value for 'apple':", ht.get_val('apple'))

ht.set_val('apple', 50)
print("Updated Hash Table:", ht)

ht.delete_val('banana')
print("After Deletion:", ht)

print("Value for 'banana':", ht.get_val('banana'))
```
