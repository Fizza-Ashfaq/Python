```python
employees = [
    {
        "id": 1,
        "name": "Ali",
        "department": "IT",
        "salary": 75000
    },
    {
        "id": 2,
        "name": "Sara",
        "department": "HR",
        "salary": 65000
    }
]

def total_salary(employee):
    total=0
    for employee in employees:
        total=total+employee["salary"]
    return total

def it_employee_salary():
    it_employee=[]
    length=0
    for employee in employees:
        if employee["department"]=="IT":
            it_employee+=employee
            length+=1
    avg=total_salary(it_employee)
    avg=avg/length
    return avg

print(f"IT department employees avg salary: {it_employee_salary()}")

print(f"Employee having salary greater than 70000: ")
for employee in employees:
    if employee["salary"]>70000:
        print(f"{employee["name"]} - {employee["salary"]}")

print("Employee working in IT department:")
for employee in employees:
    if employee["department"]=="IT":
        print(f"{employee["name"]} - {employee["department"]}")

for employee in employees:
    if employee["id"]==2:
        employee["salary"]=employee["salary"]+(employee["salary"]/10)
        print(f"Employe with id 2 salary increased: {employee["salary"]}")
        employees.remove(employee)
        print("Employee with id 2 removed successfuly")
        print(f"Updated record: {employees}")
        
```
