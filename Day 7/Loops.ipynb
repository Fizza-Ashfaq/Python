```python
## Even Positioned Characters

word=input("Enter a word")
i=0
while i<len(word):
    if i%2==0:
        print(word[i])
    i+=1
```

```python
## Jumping through While - Python

num=int(input("Enter a number: "))
i=1
power=i**2
while power <num:
    print(power)
    i+=1
    power=i**2
```

```python

## Print employee names only
employees = [
    {"id": 1, "name": "Ali", "salary": 75000},
    {"id": 2, "name": "Sara", "salary": 85000},
    {"id": 3, "name": "Ahmed", "salary": 65000}
]
print("Employees name:")
for employee in employees:
    print(employee["name"])
```

```python
## Printing tables

table=int(input("Enter a number (1-10): "))

for i in range(1,table):
    for j in range(1,10):
        print(f"{i} * {j} = {i*j}")
```

```python
## Printing negative and positive numbers

flag=True
while flag:
    num=int(input("Enter a number"))
    if num>0:
        print("Positive number")
    elif num<0:
        print("Negative number")
    else:
        print("Program ended")
        flag=False
```

```python
## skipping salaries using continue

employees = [
    {"name": "Ali", "salary": 75000},
    {"name": "Sara", "salary": 45000},
    {"name": "Ahmed", "salary": 80000},
    {"name": "Zain", "salary": 50000}
]

print("Employee salaries below 60000: ")
for employee in employees:
    if employee["salary"] <60000:
        continue
    print(employee["salary"])
```

```python
## printing using nested loops

for j in range(1,6):
    for i in range(j):
        print("*", end="")
    print()
```

```python
## List comprehensions

numbers = [12, 5, 8, 21, 30, 17, 4, 9]
new_numbers=[num for num in numbers if num%2==0 and num>10]
print(numbers)
print(new_numbers)
```

```python
logs = [
    "INFO: User 101 logged in",
    "ERROR: Database connection failed",
    "INFO: User 102 logged in",
    "WARNING: Disk space low",
    "ERROR: Authentication failed",
    "INFO: User 103 logged in",
    "ERROR: Database connection failed"
]

warning_count=0
error_count=0
info_count=0

for log in logs:
    if "INFO" in log:
        info_count+=1
    elif "ERROR" in log:
        error_count+=1
    elif "WARNING" in log:
        warning_count+=1

print(f"Number of INFO logs: {info_count}")
print(f"Number of ERROR logs: {error_count}")
print(f"Number of WARNING logs: {warning_count}")

error_msgs=[]
for log in logs:
    if "ERROR" in log and not log in error_msgs:
        error_msgs.append(log.split(":")[1].strip())
    
print(f"Error msgs: {error_msgs}")
```
