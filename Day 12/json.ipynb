```python
import json

student={
    "id": 1,
    "name": "Fizza",
    "grade": "A"
}

with open("notes.json","w") as file:
    json.dump(student,file,indent=6)

with open("notes.json","r") as file:
    content=json.load(file)

print(content)
```

```python
import json

students=[
    {"id": 1, "name": "Ali", "marks": 85},
    {"id": 2, "name": "Ahmed", "marks": 72},
    {"id": 3, "name": "Sara", "marks": 91}
]


with open("students.json","w") as file:
    json.dump(students,file,indent=4)

with open("students.json","r") as file:
    data=json.load(file)

print(data)
for student in data:
    if student["name"]=="Sara":
        print(f"Sara's marks: {student["marks"]}")
```
