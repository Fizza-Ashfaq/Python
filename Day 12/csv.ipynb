```python
import csv

with open("students.csv","w",newline="") as file:
    csvwriter=csv.writer(file)
    csvwriter.writerow(["id","name","marks"])
    csvwriter.writerow([1,"Ali",85])
    csvwriter.writerow([2,"Ahmed",72])
    csvwriter.writerow([3,"Sara",91])

with open("students.csv","r") as file:
    contents=csv.reader(file)
    for row in contents:
        print(row)
```
