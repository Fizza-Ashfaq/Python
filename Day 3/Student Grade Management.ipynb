```python
students_record={}
name=input("Enter student name: ")
subjects=int(input("Enter number of subjects: "))
students_record[name]={}
i=0

while i<subjects:
    subj_name=input(f"Enter name of subject{i+1}: ")
    marks=int(input(f"Enter marks of {subj_name}"))
    if 0<=marks<=100 :
       students_record[name][subj_name]=marks
    else:
        print("Please enter valid marks!")
    i+=1


def cal_highest():
    for name in students_record:
        highest=0
        for subject in students_record[name]:
            if highest<students_record[name][subject]:
                highest=students_record[name][subject]

    return highest


def cal_total():
    total=0
    for name in students_record:
        for subject in students_record[name]:
            total=total+students_record[name][subject]
    return total

def cal_lowest():
    lowest=list(list(students_record[name][subjects]).values())[0]
    for name in students_record:
        for subject in students_record[name]:
            if lowest>students_record[name][subject]:
                lowest=students_record[name][subject]
    return lowest

def cal_grade(grade):
    if(grade>90):
        print("Grade: A")
    elif(grade>80):
        print("Grade: B")
    elif(grade>70):
        print("Grade: C")
    elif(grade>60):
        print("Grade: D in sub")
    else:
        print("Grade: F")


print(f"Your total marks are: {cal_total()}")
print(f"Your highest marks are: {cal_highest()}")
print(f"Your lowest marks are: {cal_lowest()}")
cal_grade(cal_total())
    
```
