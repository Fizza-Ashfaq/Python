```python
year=int(input("Enter a year: "))
if (year%4==0):
    print("It's a leap year!")
else:
    print("It's not a leap year!")
```

```python
user_record={
    "user1":{
        "username" : "admin",
        "password" : "Admin@123",
        "is_active" : True,
        "is_verified" : True
    },
    "user2":{
        "username":"fizza",
        "password":"fizza0849@gmail.com",
        "is_active": True,
        "is_verified": False
    },
    "user3":{
        "username":"saba",
        "password":"saba@gmail.com",
        "is_active": False,
        "is_verified": False
    }
}

name=input("Enter name: ")
password=input("Enter password: ")
for user in user_record.values():
  if user["username"] ==name :
    if user["password"]!=password:
       print("Invalid password")
    elif user["is_active"] !=True:
        print("Account is inactive")
    elif user["is_verified"] !=True:
        print("Account verification is required")
    else:
        print("Account logged in successfully")
```
