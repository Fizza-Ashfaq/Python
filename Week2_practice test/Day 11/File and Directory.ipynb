```python
from pathlib import Path

dir_path=Path(input("Enter directory path: "))

if dir_path.is_dir():
    print(f"Directory exists")
else:
    print(f"Not found!")

file_count=0
dir_count=0

print("Files: ")
for subfile in dir_path.iterdir():
    if subfile.is_file():
        file_count+=1
        print(subfile)
    elif subfile.is_dir():
        dir_count+=1


print(f"Number of files: {file_count}")
print(f"Number of Directories: {dir_count}")
```
