```python
## common values in list

intersected_list=[]
list1=[1,2,3,4,3,4,5,6,6,7,8]
list2=[3,4,9,2]
for i in list1:
    if not i in intersected_list and i in list2:
        intersected_list.append(i)
    i+=1
print(intersected_list)
```

```python
## unCommon values in list

diff_list=[]
list1=[1,2,3,4,3,4,5,6,6,7,8]
list2=[3,4,9,2]
for i in list1:
    if not i in list2 and not i in diff_list:
        diff_list.append(i)
    i+=1
print(diff_list)
```

```python
## calculate unique categories and total amount per category

transactions = [
    {"id": 1, "type": "expense", "amount": 500, "category": "Food"},
    {"id": 2, "type": "expense", "amount": 1200, "category": "Transport"},
    {"id": 3, "type": "income", "amount": 50000, "category": "Salary"},
    {"id": 4, "type": "expense", "amount": 300, "category": "Food"},
    {"id": 5, "type": "expense", "amount": 800, "category": "Shopping"},
    {"id": 6, "type": "expense", "amount": 700, "category": "Transport"},
    {"id": 7, "type": "expense", "amount": 450, "category": "Food"},
    {"id": 8, "type": "expense", "amount": 1500, "category": "Shopping"},
]
def find_unique():
    unique=[]
    for transaction1 in transactions:
        is_unique=True
        count=0
        for transaction2 in transactions:
            if transaction1["category"]==transaction2["category"]:
                count+=1
            if count >1: 
                is_unique=False
                break
        if is_unique ==True:
            unique.append(transaction1)
    return unique

unique_transactions=find_unique()
print(f"Unique categories are: {unique_transactions}")

def total():
    total = {}

    for transaction1 in transactions:
        category = transaction1["category"]

        if category not in total:
            total[category] = 0

        total[category] += transaction1["amount"]

    return total
    
print(f"Total amount of transactions: {total()}")
```

```python
## sets

setA={1,2,'a','s',89}
setB={4,5,6,'Y','F',0,'a',1}

union_of_sets=setA.union(setB)
intersection_of_sets=setA.intersection(setB)
a_minus_b=setA.difference(setB)
b_minus_a=setB.difference(setA)

print(f"UNION OF SETS: {union_of_sets}")
print(f"UNION OF SETS: {intersection_of_sets}")
print(f"UNION OF SETS: {a_minus_b}")
print(f"UNION OF SETS: {b_minus_a}")
```

```python
##Highest and lowest priced product

products = [
    ("Laptop", 120000),
    ("Mouse", 2500),
    ("Keyboard", 5000)
]

highest=products[0]
for product in products:
    if highest[1]<product[1]:
        highest=product

print(f"Highest priced product: {highest}")

lowest=products[0]
for product in products:
    if lowest[1]>product[1]:
        lowest=product

print(f"Lowest priced product: {lowest}")
```
