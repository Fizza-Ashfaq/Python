```python
import re
students_record={}

def create_record():
    id=len(students_record)+1
    subjects=()
    students_record[id]={}
    students_record[id]["name"]=input("Enter name: ")
    students_record[id]["age"]=int(input("Enter age: "))
    students_record[id]["email"]=input("Enter email: ")
    sub=input("Enter subjects (e.g math urdu eng): ")

    subjects=tuple(re.findall(r'\w+', sub))
    students_record[id]["subjects"]=subjects
    students_record[id]["marks"]={}
    
    for subject in subjects:
        students_record[id]["marks"][subject]=int(input(f"Enter marks of {subject}"))
    students_record[id]["attendance"]=int(input("Enter attendance: "))

def view_record():
    print(students_record)

def search_record():
    to_search=int(input("Enter id to search: "))
    if to_search in students_record:
            print(students_record[to_search])
    else:
        print("id doesn't exist")

def update_record():
    to_update=int(input("Enter student id to update: "))
    if to_update in students_record:
        attribute=input("Enter attribute to update")
        students_record[to_update][attribute]=input(f"Enter {attribute}: ")
        print("Record updated successfully")
    else:
        print("Record doesn't exist")

def delete_record():
    to_delete=int(input("Enter student id to delete: "))
    if to_delete in students_record:
        students_record.pop(to_delete)
        print("Record deleted successfully")
    else:
        print("Record doesn't exist")

def total_marks():
    total=0
    for student in students_record:
        for subject in students_record[student]["marks"]:
            total=total+students_record[student]["marks"][subject]
    print(f"Total marks of students: {total}")
    return total

def avg_marks():
    total=total_marks()
    for student in students_record:
        length+=len(students_record[student]["subjects"])
    avg=total/length
    print(f"Average of students marks: {avg}")

cont=True
while cont:
    print("----Student Record Management System----")
    print("1. Enter a record")
    print("2. Search a record")
    print("3. View records")
    print("4. Update a record")
    print("5. Delete a record")
    print("6. Total marks of students")
    print("7. Average of students")
    print("8. Exit")
    op=input(f"Choose an option(1-8): \n\n")

    match op:
        case "1":
            create_record()
        case "2":
            search_record()
        case "3":
            view_record()
        case "4":
            update_record()
        case "5":
            delete_record()
        case "6":
            total_marks()
        case "7":
            avg_marks()
        case "8":
            cont=False
        case _:
            print("Please enter a valid record ")
```
