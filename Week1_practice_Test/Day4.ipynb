```python
##1. String Information

string=input("Enter a string: ")
print(f"Length of  string : {len(string)}")
print(f"First character: {string[0]}")
print(f"Last character: {string[-1]}")
print(f"String in uppercase: {string.upper()}")
print(f"String in lowercase: {string.lower()}")
```

```python
##2. Reverse a String
string=input("Enter a string: ")
print(f"String in Reversed order: {string[::-1]}")
```

```python
##3. Count Characters

string=input("Enter a string: ")

print(f"Number of characters: {len(string.replace(" ",""))}")
print(f"Number of spaces: {(string.count(" "))}")
count=0
for i in string :
    if i=="a" or i=="e" or i=="i" or i=="o" or i=="u":
        count+=1
print(f"Number of vowels: {count}")
```

```python
##4. Username Validator

username=input("Enter name: ")

if len(username)<=4:
    print("Name length must be atleast 5 ")
if username.find(" ")>=0:
    print("Name cannot contain spaces")
if not username[0].isalpha():
    print("Name must start with an alphabet")

```

```python
##5. Palindrome Checker

string=(input("Enter a string: ").replace(" ","")).lower()
rev_string=string[::-1]

if string==rev_string:
    print("Its a palindrome")
else:
    print("Its not a palindrome")
```

```python
##6. Password Strength Validator

password=input("Enter password: ")

if len(password)<8:
    print("Password length must be atleast 8 characters")
if not any(i.isupper() for i in password):
    print("Password must contain atleast one upper case letter")
if not any(i.lower() for i in password):
    print("Password must contain atleast one lower case letter")
if not any(i.isdigit() for i in password):
    print("Password must contain atleast one digit (e.g 1,2..) ")
if not any(not i.isalnum() for i in password):
    print("Password must contain atleast one special character (e.g @,&..)")
```
