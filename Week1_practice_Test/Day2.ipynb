```python
##1. Type Identification

string="Python"
integer=25
flt=3.14
flag=True
none=None

print(f"{string}: {type(string)}")
print(f"{integer}: {type(integer)}")
print(f"{flt}: {type(flt)}")
print(f"{flag}: {type(flag)}")
print(f"{none}: {type(none)}")
```

```python
##2. Marks Calculator

subj1=int(input("Enter marks of subject 1: "))
subj2=int(input("Enter marks of subject 2: "))
subj3=int(input("Enter marks of subject 3: "))

total=subj1+subj2+subj3
avg=(total/3)


print(f"Total marks: {total}")
print(f"Average of marks: {avg}")
```

```python
##3. Type Conversion

weight=float(input("Enter your weight: "))
height=float(input("Enter your height in meters: "))
BMI=weight/height**2
print(f"Your BMI is: {BMI}")
```

```python
##4. Operator Challenge

num1=int(input("Enter number 1:"))
num2=int(input("Enter number 2:"))

print(f"Sum: {num1+num2}")    
print(f"Difference: {num1-num2}")
print(f"Product: {num1*num2}")
print(f"Quotiont: {num1/num2}")
print(f"Floor Division: {num1//num2}")
print(f"Remainder: {num1%num2}")
print(f"{num1} raise to the power of {num2}: {num1**num2}")
```
