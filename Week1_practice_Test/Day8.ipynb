```python
##2. Calculator Functions

def add(num1,num2):
    return num1+num2

def subtract(num1,num2):
    return num1-num2

def multiply(num1,num2):
    return num1*num2

def divide(num1,num2):
    return num1/num2


num1=int(input("Enter number 1: "))
num2=int(input("Enter number 2: "))

print(f"Addition: {add(num1,num2)}")
print(f"Subtraction: {subtract(num1,num2):}")
print(f"Multiplication: {multiply(num1,num2)}")
print(f"Division: {divide(num1,num2)}")
```

```python
##4. Default Arguments

def calculate_discount(price, discount=10):
    return price-discount


print(f"Without providing discount: {calculate_discount(900)}")
print(f"With providing discount: {calculate_discount(900,20)}")
```

```python
##4. Expense Function

expenses=[12,34,56,789,78]

def expense_function(expense):
    total=0
    for expense in expenses:
        total=total+expense

    length=len(expenses)
    avg=total/length
    expenses.sort()
    highest=expenses[-1]
    lowest=expenses[0]

    result={"total":total,"avg":avg,"highest":highest,"lowest":lowest}
    return result

# returned_values=[]
returned_values=expense_function(expenses)
print(f"Total expense: {returned_values["total"]}")
print(f"Average expense: {returned_values["avg"]}")
print(f"Highest expense: {returned_values["highest"]}")
print(f"Lowest expense: {returned_values["lowest"]}")
```

```python
##6. Expense Tracker Refactoring Challenge

expenses=[]

def add_expense():
    expense=int(input("Enter expenses: "))
    expenses.append(expense)
    print("Expenses added successfuly!")
    print(f"Expenses: {expenses}")

def view_expenses():
    print(f"Expenses: {expenses}")

def calculate_total():
    total=0
    for expense in expenses:
        total+=expense
    print(f"Total expenses: {total}")

def calculate_average():
    total=0
    for expense in expenses:
        total+=expense
    length=len(expenses)
    avg=total/length
    print(f"Average: {avg}")

def delete_expense():
    to_delete=int(input("Enter expense to delete:"))
    for expense in expenses:
        print(expense)
        if expense==to_delete:
            expenses.remove(to_delete)
            print("Expense removed successfully")

flag=True
while flag:
    print("\n\nExpense Tracker System")
    print("1. Add Expense \n2. View Expenses \n3. Total Expenses \n4. Average Expense \n5. Delete Expense \n6. Exit")
    op=input("Choose an option(1-6):")

    match op:
        case "1":
            add_expense()
        case "2":
            view_expenses()
        case "3": 
            calculate_total()
        case "4":
            calculate_average()
        case "5":
            delete_expense()
        case "6":
            flag=False
            print("Exiting")
        case _:
            print("Please enter a valid operation")
```
