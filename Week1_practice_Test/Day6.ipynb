```python
##2. Dictionary Methods

students = {
    "name": "Ali",
    "age": 22,
    "grade": "A"
}

print(f"Keys: {students.keys()}")
print(f"Vallues: {students.values()}")

for key,value in students.items():
    print(f"{key}: {value}")

print(f"Getting name through get method: {students.get("name")}")
```

```python
##3. Employee Search

employees={
    1:
    {"id":1,"name":"Ali"},
    2:
    {"id":2,"name":"Ahmad"},
    3:
    {"id":3,"name":"Arham"}
}

id=int(input("Enter employee id to search: "))

flag=False
for employee in employees:
        if id==employees[employee]["id"]:
            print(employees[employee])
            print("Employee found!")
            flag=True
if flag==False:
    print("Employee not found")
```

```python
##4. Word Frequency Counter

sentence="python is easy and python is powerful"
list_of_words={}
for sen1 in sentence:
    count=0
    if sen1 in list_of_words:
        break
    else:
        for sen2 in sentence:
            if sen1==sen2:
                count+=1
    list_of_words[sen1]=count

print(list_of_words)
```

```python
##5. Nested Employee Records

employees = {
    101: {
        "name": "Ali",
        "department": "IT",
        "salary": 80000
    },
    102: {
        "name": "Sara",
        "department": "HR",
        "salary": 70000
    }
}

to_search=int(input("Enter employee id to search: "))

flag=True
for employee in employees:
    if to_search==employee:
        flag==False
        print("Employee found!")
        print(employee)
        updated_salary=int(input("Enter new salary to update: "))

        employees[employee]["salary"]=updated_salary
        print("Employee salary updated successfuly")
        
        updated_dep=input("Enter new department to update: ")
        employees[employee]["department"]=updated_dep
        print("Employee department updated successfuly")
        break

if flag==True:
        print("Employee not found")


flag=True
to_delete=int(input("Enter employee id to delete: "))
for employee in employees:
    if to_delete==employee:
        flag=False
        employees.pop(to_delete)
        print("Employee Deleted sucessfully!")
        break

if flag==True:
        print("Employee not found")
```

```python
##6. Highest Paid Employee

employees = [
    {"name": "Ali", "salary": 80000},
    {"name": "Sara", "salary": 95000},
    {"name": "Ahmed", "salary": 70000},
    {"name": "John", "salary": 90000}
]

highest=employees[0]["salary"]
for employee in employees:
    if highest<employee["salary"] :
        highest=employee["salary"] 

print(f"Highest salary: {highest}")
```
