```python
##1.  Positive, Negative or Zero

num=int(input("Enter a number: "))
if num>0:
    print("Positive number")
elif num<0:
    print("Negative number")
else:
    print("Zero")
```

```python
##2. Even or Odd

num=int(input("Enter a number: "))
if num%2==0:
    print("Its an even number")
else:
    print("its an odd number")
```

```python
##3. Student Grade

per=int(input("Enter your percentage: "))
if 100>=per>=90 :
    print("Grade A+")
elif 90>per>=80 :
    print("Grade A")
elif 80>per>=70 :
    print("Grade B")
elif 70>per>=60 :
    print("Grade C")
elif 60>per>=50 :
    print("Grade D")
elif 50>per:
    print("Grade F")
else:
    print("Invalid Input")
```

```python
##5. Leap Year Challenge

year=int(input("Enter a year: "))
if(year%4==0 and not year%100==0):
    print("Its a leap year ")
else:
    print("Its not a leap year ")
```

```python
##6. Student Grade Management

subjects=[]
i=1
while i<=5:
    subjects.append(int(input(f"Enter marks of subject {i} (0-100):")))
    i+=1

fail=False
for subject in subjects:
    if subject<=33:
        fail=True
        break

if fail==True:
    print("Result: F \nReason: Failed one or more subjects")
else:
    total=0
    for subject in subjects:
        total+=subject
    avg=total/len(subjects)
    if 100>=avg>=90:
        print("Grade: A+")
    elif 90>avg>=80:
        print("Grade: A")
    elif 80>avg>=70:
        print("Grade: B")
    elif 70>avg>=60:
        print("Grade: C")
    elif 60>avg>=50:
        print("Grade: D")
    elif 50>avg:
        print("Grade F")
    else:
        print("Please enter marks within the valid range (i.e. 0-100) ")
```
