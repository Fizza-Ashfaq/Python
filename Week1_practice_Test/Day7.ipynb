```python
##1. Numbers 1–100

for i in range(0,100):
    i+=1
    print(i)
    
```

```python
##2. Multiplication Table

num=int(input("Enter a number(1-10): "))

if 0<num<=10:
    for i in range(1,num+1):
        print(f"\nTable of {i}")
        for j in range(1,11):
            print(f"{i}*{j}={i*j}")
else:
    print("Please enter number in range 1-10")
```

```python
##3. Sum of Numbers

num=int(input(f"Enter a number: "))
sum=0
for i in range(1,num+1):
    sum+=i
print(f"Sum: {sum}")
```

```python
##4. Skip Multiples

for i in range(1,100):
    if i%3==0:
        continue
    elif i==80:
        break
    else:
        print(i)

```

```python
##5. Pattern Challenge

for i in range(0,5):
    for j in range(0,i+1):
        print("*",end="")
    print("")

## Modified program
print("Modified program: ")
i=5
for i in range(0,5):
    j=5
    while j>i:
        print("*",end="")
        j=j-1
    print("")
```
