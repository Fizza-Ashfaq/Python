```python
##1. List Operations

student_names=['Ali','Ahmad','Saba','Abiha','Minahil']

student_names.append('Arham')
print(f"List after appending: {student_names}")
student_names.remove(student_names[0])
print(f"List after removing first name: {student_names}")
student_names.sort()
print(f"List after sorting: {student_names}")
student_names.reverse()
print(f"List after reversing: {student_names}")
```

```python
##2. Tuple Practice

personal_info=('fizza',22,'Lahore')
print(f"Name: {personal_info[0]}")
print(f"Age: {personal_info[1]}")
print(f"City: {personal_info[2]}")
```

```python
##3. Remove Duplicates

numbers = [1, 2, 3, 2, 4, 5, 3, 6, 1]
new_set=set(numbers)
print(f"Set: {new_set}")
```

```python
##4. Set Operations


python_students = {"Ali", "Ahmed", "Sara", "John"}
java_students = {"Ahmed", "John", "David", "Sara"}

print(f"Students studiying both python and java: {python_students.intersection(java_students)}")
print(f"Students studiying only python: {python_students.difference(java_students)}")
print(f"Students studiying only java: {java_students.difference(python_students)}")
```

```python
##5. Second Largest Number

numbers = [10, 5, 20, 20, 60, 15, 10]
unique_numbers=set(numbers)
largest=0
second_largest=0
for number in unique_numbers:
    if largest<number:
        largest=number
for number in unique_numbers:
    if second_largest<number and number<largest:
        second_largest=number
print(f"Second largest number: {second_largest}")
```

```python
##6. Student Record System

students = [
    ["Ali", 85],
    ["Ahmed", 72],
    ["Sara", 91],
    ["John", 65]
]

print(f"All students: {students}")
highest=0
for student in students:
    if student[1]>highest:
        highest=student[1]
print(f"Highest scorer: {highest}")

lowest=highest
for student in students:
    if student[1]<lowest:
        lowest=student[1]

print(f"Lowest scorer: {lowest}")

total=0
for student in students:
    total=total+student[1]
avg=total/len(students)
print(f"Average marks: {avg}")
```
