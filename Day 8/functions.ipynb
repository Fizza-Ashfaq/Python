```python
def calculate_expense(total, tax=0):
    total=total+tax
    return total

print(calculate_expense(1000,987))
```

```python

def create_employee(name, department, salary=50000):
    employee={
        "name":name,
        "department":department,
        "salary":salary
    }
    return employee

employee = create_employee(
    salary=75000,
    name="Ali",
    department="IT"
)
print(employee)
```

```python
def calculate_total(*expenses):
    total=0
    i=0
    while i<len(expenses):
        total=total+expenses[i]
        i+=1
    return total

print(calculate_total(500, 200, 150))
print(calculate_total(100, 50, 75, 25))
```

```python

def create_employee(**details):
    employee={}
    for key in details:
        employee[key]=details[key]
    return employee

employee = create_employee(
    name="Ali",
    department="IT",
    salary=75000,
    experience=2
)

print(employee)
```
