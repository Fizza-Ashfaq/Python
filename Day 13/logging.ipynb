```python
## Logging configuration

import logging

logging.basicConfig(
    filename="inventory.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)
```

```python
## Inventory Audit Logging

logging.info("Inventory system started")

p_id=int(input("Enter product id: "))
name=input("Enter a product name: ")
quantity=int(input("Enter product quantity: "))
product={
    "id":p_id,
    "name":name,
    "quantity":quantity
}

logging.info(f"Product added - {product["id"]}")


to_update=input("Enter attribute u want to update: ")
try:
    product[to_update]=input(f"Enter {to_update}: ")
except ValueError:
    logging.error("value error")
else:
    logging.info(f"Product updated - {product['id']}")
    print("Product updated")
```
