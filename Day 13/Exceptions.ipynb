```python
##1. Value Error

try:
    num=int(input("Enter a number: "))
    print(num)
except ValueError:
    print(f"Please enter a number")
```

```python
##2. Type error

try:
    result=2+3
except TypeError:
    print(f"Type error")
else:
    print(result)
```

```python
##3. ZeroDivision error

try:
    result=2/0
except ZeroDivisionError:
    print("ZeroDivisionError")
```

```python
##4. Index Error

try:
    list1=[1,2,3,4,4]
    print(list1[9])
except IndexError:
    print("Index out of range")
```

```python
##5. Key Error

try:
    dic={
        "id":1
    }

    print(dic["name"])
except KeyError:
    print("Key error")
```

```python
##6. File not found error

try:
    open("Exceptions & Debugging.ipynb")
except FileExistsError:
    print("FileNotFoundError")
else:
    print("File exists")
```

```python
##7. Name Error

try:
    print(whatever)
except NameError:
    print("NameError")
```

```python
##8. Raise

age=int(input("Enter your age:"))

if age<0:
    raise ValueError("Age can't be less then zero")
else:
    print(age)
```

```python
##9. Custom Exceptions

class InvalidAgeError(Exception):
    pass

try:
    age=int(input("Enter ur age: "))
    if age>100:
        raise InvalidAgeError
    else:
        print("Valid age")
except InvalidAgeError:
    print("Please type a valid age")
```

```python
##10. Invalid inventory quantity exception

class InvalidInventoryQuantity(Exception):
    pass

try:
    quantity=int(input("Enter inventory quantity: "))
    if quantity<=0:
        raise InvalidInventoryQuantity("Quantity must be greater than 0")
except InvalidInventoryQuantity as e:
    print(e)
else:
    print(f"Quantity: {quantity}")
```
