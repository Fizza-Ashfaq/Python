```python
from pathlib import Path

print(f"Current Directory: ")
for item in Path.cwd().iterdir():
    print(item)

filename=input("Enter a filename: ")

if Path(filename).exists():
    print("It exists")
    if(Path(filename).is_file()):
        print("It's a file")
    else:
        print("It's a directory")
else:
    print("File not found")
```
