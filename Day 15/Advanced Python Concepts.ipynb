```python
##1. Products iterator

products = ["Laptop", "Mouse", "Keyboard", "Monitor"]

iterator=iter(products)
while True:
    try:
        book=next(iterator)
        print(book)
    except StopIteration:
        break
```

```python
##2. Generators

def even(n):
    for num in range(n+1):
        if num%2==0:
            yield num

n=int(input("Enter a number:"))
for num in even(n):
    print(num)
```

```python
##3. yield

def numbers():
    print("A")
    yield 1
    print("B")
    yield 2
    print("C")
    yield 3


x = numbers()

print("X")
print(next(x))
print("Y")
print(next(x))
print("Z")
```

```python
##4. Decorators

def log_function(function):
    def wrapper():
        print("Function started")

        function()

        print("Function finished")

    return wrapper


@log_function
def process_payment():
    print("Payment processed")

process_payment()
```

```python
##5. Decorator with arguments

def log_function(function):
    def wrapper(*args):
        print("Function Started")

        function(*args)
        print("Function ended")
    return wrapper
@log_function
def create_user(name, age):
    print(f"User {name}, age {age} created")

create_user("fizza",22)
```

```python
##6. Context manager

class FileManager:

    def __enter__(self):
        self.file = open("test.txt", "w")
        return self.file

    def __exit__(self, exc_type, exc, tb):
        self.file.close()


with FileManager() as file:
    file.write("Hello Python")
```

```python
##8. Data class

from dataclasses import dataclass

@dataclass
class Employee:
    employee_id: int
    name: str
    department: str
    salary: float

emp=Employee(1,"Fizza","whatever",2900000.00)
print(emp)
```

```python
##10. zip

products = ["Laptop", "Mouse", "Keyboard"]
quantities = [5, 20, 12]

for product,quantity in zip(products,quantities):
    print(f"{product}: {quantity} units")
```

```python
##11. filter and map

transactions = [1500, -200, 3000, -500, 700, 0]

def positive(transactions):
    return transactions>0

def increased_price(p_transactions):
    return p_transactions+(p_transactions/100*5)

p_transactions=list(filter(positive,transactions))
print(f"Positive numbers: {p_transactions}")

updated_p_transactions=list(map(increased_price,p_transactions))
print(f"Update positive transactions: {updated_p_transactions}")
```
