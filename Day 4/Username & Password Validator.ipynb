```python
while True:
    name=(input("Enter username: ")).strip()
    if not all(char.isalnum() or char == "_" for char in name):
       print("Please enter a valid username ") 
       continue
   
    if not name[0].isalpha():
       print("username must start with a alphabet ") 
       continue
   
    if not 5<= len(name) <=20:
       print("Username must be 5-20 characters")
       continue
   
    break



while True:
    password=(input("Enter a password: ")).strip()

    if len(password)<8:
       print("Password length must be atleast 8 characters")
       continue
   
    if not any(char.islower() for char in password):
       print("Password must contain atleast one lower case letter (e.g a,b) ")
       continue
   
    if not any(char.isupper() for char in password):
       print("Password must contain atleast one upper case letter (e.g A,B) ")
       continue
   
    if not any(char.isdigit() for char in password):
       print("Password must contain atleast one digit (e.g 1,2..) ")
       continue 
   
    if not any(not char.isalnum() for char in password):
       print("Password must contain atleast one special character (e.g @,&..) ") 
       continue
   
    break

print("Login successful")
```
